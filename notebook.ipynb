### Tasks

- Identify the single feature of the data that is the best predictor of whether a customer will put in a claim (the `"outcome"` column), excluding the `"id"` column.

- Store as a DataFrame called `best_feature_df`, containing columns named `"best_feature"` and `"best_accuracy"` with the name of the feature with the highest accuracy, and the respective accuracy score.

```python
# Import required modules
import pandas as pd
import numpy as np
from statsmodels.formula.api import logit
```

```python
# importing and reading csv file

df= pd.read_csv('csv_files/car_insurance.csv')
df.head()
```

### Preparing the data

```python
df.info()
```

```python
df.isna().sum()
```

##### Here `credit_score` and `annual_mileage` have null values.

##### To fill these numerical columns we must fill it with their median

```python
# Filling numerical columns with median
credit_score_md = df['credit_score'].median()
df['credit_score'] = df['credit_score'].fillna(credit_score_md)

annual_mileage_md = df['annual_mileage'].median()
df['annual_mileage'] = df['annual_mileage'].fillna(annual_mileage_md)
```

```python
df.isna().sum()
```

##### Here the columns `driving_experience`, `education`, `income`, `vehicle_year` and `vehicle_type` are strings. They need to be encoded as string values cannot be use in regression.

```python
categorical_cols = ['driving_experience', 'education', 'income', 'vehicle_year', 'vehicle_type']

for col in categorical_cols:
    df[col] = pd.Categorical(df[col]).codes
```

```python
df.head()
```

```python
df.dtypes
```

### 3. Logistic regression

```python
features = [col for col in df.columns if col not in  {'id', 'outcome'}]

accuracies = []

for col in features:
    formula = f'outcome ~ {col}'
    model = logit(formula, data=df).fit(disp=False) 
    conf_matrix = model.pred_table()
    
    TN = conf_matrix[0][0]
    FP = conf_matrix[0][1]
    FN = conf_matrix[1][0]
    TP = conf_matrix[1][1]
    
    total = conf_matrix.sum()
    accuracy = (conf_matrix[0,0] + conf_matrix[1,1])/ total
    accuracies.append((col, accuracy))

    print(f'{col} :{accuracy}')
```

### Storing the best feature

```python
best = max(accuracies, key=lambda x: x[1])
best_feature_df = pd.DataFrame({"best_feature": [best[0]], "best_accuracy": [best[1]]})
print(best_feature_df)
```
